```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
df = pd.read_csv("crop_Yield.csv.csv")
df = df.drop_duplicates()
categorical_cols = df.select_dtypes(include=["object"]).columns
le = LabelEncoder()
for col in categorical_cols:
    df[col] = le.fit_transform(df[col].astype(str))
numerical_cols = df.select_dtypes(include=["number"]).columns


scaler = StandardScaler()

df[numerical_cols] = scaler.fit_transform(df[numerical_cols])

print(df.head())

df.to_csv("amar_prepared.csv", index=False)
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("/content/crop_Yield.csv.csv")

season_share = df.groupby("Season")["Production"].sum().sort_values(ascending=False)

print("Total Production by Season:")
print(season_share)

plt.figure(figsize=(8, 8))
plt.pie(
    season_share,
    labels=season_share.index,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Season Share by Total Crop Production")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["Annual_Rainfall"], df["Yield"], alpha=0.5)
plt.title("Annual Rainfall vs Crop Yield")
plt.xlabel("Annual Rainfall")
plt.ylabel("Crop Yield")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("crop_Yield.csv.csv")

plt.figure(figsize=(8, 5))
plt.hist(df["Yield"].dropna(), bins=20, edgecolor="black")
plt.title("Distribution of Crop Yield")
plt.xlabel("Yield")
plt.ylabel("Frequency")
plt.show()
```

```python
df.groupby("Season")["Yield"].mean().plot(kind="bar")

plt.title("Average Crop Yield by Season")
plt.xlabel("Season")
plt.ylabel("Average Yield")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
df.groupby("State")["Yield"].mean().nlargest(10).plot(kind="bar")

plt.title("Top 10 States by Average Crop Yield")
plt.xlabel("State")
plt.ylabel("Average Yield")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("crop_Yield.csv.csv")

top_crops = df.groupby("Crop")["Production"].sum().nlargest(10).sort_values()

top_crops.plot(kind="barh", figsize=(10, 6), edgecolor="black")

plt.title("Top 10 Crops by Total Production")
plt.xlabel("Total Production")
plt.ylabel("Crop")
plt.tight_layout()
plt.show()
```
